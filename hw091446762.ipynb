測試一

```python
# 1. 設定活動名稱與單價（用 dict 存放活動資訊）
prices = {
    "board": 120,
    "movie": 80
}
titles = {
    "board": "桌遊體驗",
    "movie": "電影欣賞"
}

# 2. 原始報名資料（測試一範例：請依你的實際測試輸入替換）
raw_board = "Amy, ben , AMY, cora"
raw_movie = "ben, Dora, eric, BEN"

# 3. 先轉小寫、去頭尾半形空白，再以逗號切分成 list（保留重複）
board_list = raw_board.lower().strip().replace(" ", "").split(",")
movie_list = raw_movie.lower().strip().replace(" ", "").split(",")

# 4. 用 set 去重，並轉回排序後的 list 方便檢視
board_set = set(board_list)
movie_set = set(movie_list)

board_unique = sorted(board_set)
movie_unique = sorted(movie_set)

# 5. 計算各場人數、交集、聯集與兩個方向的差集
board_count = len(board_unique)
movie_count = len(movie_unique)

both_list = sorted(board_set & movie_set)          # 交集：共同參加者
only_board_list = sorted(board_set - movie_set)    # 差集 1：只參加桌遊者
only_movie_list = sorted(movie_set - board_set)    # 差集 2：只參加電影者
all_unique_list = sorted(board_set | movie_set)    # 聯集：全部參加者

both_count = len(both_list)
total_unique_count = len(all_unique_list)
total_entries = board_count + movie_count          # 兩場分別計費的報名人次

# 6. 通知名單前兩位（從全部不重複名單取切片）
notify_top2 = all_unique_list[:2]

# 7. 計算應收金額
board_fee = board_count * prices["board"]
movie_fee = movie_count * prices["movie"]
total_fee = board_fee + movie_fee

# 8. 使用 f-string 輸出摘要
print("HW09 活動報名摘要")
print(f"活動：{titles['board']}，每人 {prices['board']:.2f} 元")
print(f"活動：{titles['movie']}，每人 {prices['movie']:.2f} 元")
print(f"桌遊整理串列：{board_list}")
print(f"電影整理串列：{movie_list}")
print(f"桌遊去重名單：{board_unique}")
print(f"電影去重名單：{movie_unique}")
print(f"桌遊人數：{board_count} 人")
print(f"電影人數：{movie_count} 人")
print(f"共同參加者：{both_list}")
print(f"共同人數：{both_count} 人")
print(f"只參加桌遊者：{only_board_list}")
print(f"只參加電影者：{only_movie_list}")
print(f"全部參加者：{all_unique_list}")
print(f"總不重複人數：{total_unique_count} 人")
print(f"報名人次：{total_entries} 人次")
print(f"通知名單前兩位：{notify_top2}")
print(f"桌遊應收：{board_fee:.2f} 元")
print(f"電影應收：{movie_fee:.2f} 元")
print(f"總應收：{total_fee:.2f} 元")
```
